```python
import pandas as pd
import plotly.graph_objects as go
```

# 퍼널 차트 시각화

```python
# 1. 로그 데이터 로드 및 퍼널 집계
df = pd.read_csv('delivery_app_logs.csv')
step_order = ['1_app_opened', '2_home_viewed', '3_cart_interacted', '4_purchase_completed']
step_labels = ['1. 진입 (App Opened)', '2. 탐색 (Home Viewed)', '3. 장바구니 (Cart)', '4. 구매 완료 (Purchase)']

funnel_counts = [df[df['step'] == s]['user_id'].nunique() for s in step_order]

# 2. 배달특급 메인 블루 그라데이션 (위에서 아래로 갈수록 진해짐)
custom_colors = [
    '#AECDF8',  # 1단계 (가장 연한 블루)
    '#679DF5',  # 2단계
    '#1D68F2',  # 3단계 (배달특급 앱 아이콘 메인 컬러)
    '#0F3B99'   # 4단계 (가장 진한 딥 블루)
]

# 3. Plotly 퍼널 차트 생성
fig = go.Figure(go.Funnel(
    y=step_labels,
    x=funnel_counts,
    textinfo="value+percent initial",
    marker=dict(color=custom_colors)
))

fig.update_layout(
    title="<b>[배달특급] 핵심 퍼널 분석 및 이탈 구간 시각화</b>",
    template="plotly_white",
    font=dict(family="AppleGothic", size=12)
)

# HTML 파일로 저장 및 브라우저 확인
fig.write_html("funnel_chart.html")
print("Plotly 인터랙티브 퍼널 차트('funnel_chart.html') 생성 완료!")
```

# A/B테스트 퍼널 시각화

```python
# ==========================================
# 📊 [함수 3] A/B 테스트 비교 '퍼널' 차트 생성 함수
# ==========================================
def get_ab_test_funnel_chart(data):
    # 그룹별 데이터 집계
    group_a = data[data['ui_group'] == 'A(기존 UI)'].groupby('step')['user_id'].nunique().reindex(step_order)
    group_b = data[data['ui_group'] == 'B(개선 UI)'].groupby('step')['user_id'].nunique().reindex(step_order)

    fig = go.Figure()

    # 기존 UI (A그룹) - 회색
    fig.add_trace(go.Funnel(
        name='A (기존 UI)',
        y=step_labels,
        x=group_a.values,
        textinfo="value+percent initial",
        marker=dict(color="#B0BEC5")
    ))
    
    # 개선 UI (B그룹) - 배달특급 블루
    fig.add_trace(go.Funnel(
        name='B (개선 UI - 선탐색/퀵필터)',
        y=step_labels,
        x=group_b.values,
        textinfo="value+percent initial",
        marker=dict(color="#1D68F2")
    ))

    # 레이아웃 설정
    fig.update_layout(
        title="<b>A/B 테스트 그룹별 퍼널 전환 비교</b>",
        template="plotly_white",
        legend=dict(x=0.7, y=0.9), # 범례 위치 조정
        font=dict(family="AppleGothic", size=12) # 갤럭시북 윈도우 환경에 맞춘 맑은 고딕
    )
    
    return fig

# 함수 실행 및 결과물 저장
ab_funnel_fig = get_ab_test_funnel_chart(df)
ab_funnel_fig.write_html("ab_test_funnel_interactive.html")
ab_funnel_fig.show()

print("✅ A/B 테스트 Plotly 퍼널 차트 저장 완료!")
```
